```python
## variables Length Arguments

## Positional arguments

def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,"shanky")
```

```python
##  keywords arguments
 
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="shanky", age="24", number="909090909")
```

```python
## Example: temperature conversion

def convert_temperature(temp, unit):
    if unit == 'C':
        return temp * 9/5 +32   ## Celsius to Fahrenhiet
    elif unit == "F":
        return (temp-32) * 5/9  ## Frahenhiet to Celsius
    else:
        return None
    
    

```

```python
print(convert_temperature(32, 'C'))
print(convert_temperature(89.6, 'F'))
```

```python
## Example: Password Strength Checker 

def is_trong_password(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True
    

## calling the function
is_trong_password("shdggsg947#SNBD")
```

```python
## Example: Calculating the total cost in shopping cart

def total_cost(cart):
    init_cost = 0
    for item in cart:
        init_cost+=item['price']*item['quantity']

    return init_cost



## Example of cart data 
cart=[
    {'name':'Apple','price':0.5, 'quantity':4},
    {'name':'Banana','price':0.3, 'quantity':6},
    {'name':'Orange','price':0.7, 'quantity':3}
]

## calling the function

total = total_cost(cart)
print(total)
```
